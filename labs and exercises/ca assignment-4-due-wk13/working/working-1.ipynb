```python
# going to create a dictionary of column name and label name pai mappings, label name is more intuitive and human readable
# create a list to store the columnname and label name values whic will read from the downloaded csv
col_label_text = []
# open the csv and read lines which start with # and append to the list we created
with open(file_path, "r", encoding='ISO 8859-1') as f:
    for line in f:
        if line.startswith("# COLUMN"):
            col_label_text.append(line.strip())

#have a quick lookm make sure all good 
print(len(col_label_text))
print(col_label_text)

# create dictinoary to store column name label name pairs
col_label_dict ={}
# using regex to match STRING SBEGINING WITH "# COLUMN" then a >= 1 whitespace, >=1non-whtespace character, colon :, >= 1 spaces then any character zero or more, everythgin after that
pattern = r"# COLUMN\s+(\S+):\s+(.*)"

for line in col_label_text:
    m = re.match(pattern, line)
    if m:
        col, label = m.groups()
        col_label_dict[col] = label
        
print("Extracted mappings:", len(col_label_dict))
print(list(col_label_dict.items())[:10])
```

```python
from astroquery.gaia import Gaia

# --- Set TAP Server to a reliable mirror (e.g., VizieR) ---
Gaia.tap_server = "http://tapvizier.u-strasbg.fr/TAPVizieR/tap"

# --- 1. Query the Astrophysical Parameters (Teff, logg, [M/H]) ---
print("Querying the full Astrophysical Parameters table...")
ap_query = "SELECT * FROM gaiadr3.ap_gsp_spec"
ap_job = Gaia.launch_job_async(ap_query)
ap_results = ap_job.get_results()
ap_results.write('gaia_dr3_gsp_spec_full.csv', format='csv', overwrite=True)
print(f"Downloaded {len(ap_results)} rows of Astrophysical Parameters.")

# --- 2. Query the Evolutionary Parameters (Mass, Radius, Age) ---
print("Querying the full Evolutionary Parameters table...")
flame_query = "SELECT * FROM gaiadr3.flame_legacy"
flame_job = Gaia.launch_job_async(flame_query)
flame_results = flame_job.get_results()
flame_results.write('gaia_dr3_flame_legacy_full.csv', format='csv', overwrite=True)
print(f"Downloaded {len(flame_results)} rows of Evolutionary Parameters.")

# --- 3. Astrometry Table (Critical, but must be filtered due to size) ---
# NOTE: The full gaia_source is too large for TAP download.
# We apply a conservative magnitude filter (G < 18) to get a manageable subset
# of the brightest, highest-precision stars.
print("Querying a bright subset of the core Astrometry table...")
astrometry_query = """
    SELECT source_id, ra, dec, parallax, pmra, pmdec, phot_g_mean_mag 
    FROM gaiadr3.gaia_source 
    WHERE phot_g_mean_mag < 18.0
    AND parallax_over_error > 5
"""
astrometry_job = Gaia.launch_job_async(astrometry_query)
astrometry_results = astrometry_job.get_results()
astrometry_results.write('gaia_dr3_astrometry_bright.csv', format='csv', overwrite=True)
print(f"Downloaded {len(astrometry_results)} rows of Astrometry Data.")
```
