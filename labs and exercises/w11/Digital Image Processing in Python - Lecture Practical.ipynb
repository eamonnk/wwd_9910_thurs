### Task 7

Modify the code belo and observe the results for different numbers of segments.

Remove the `kind` parameter from `label2rgb`. What happens to the output image?


````python
from skimage.segmentation import slic
from skimage.color import label2rgb
from skimage import io
import numpy as np
import matplotlib.pyplot as plt

# Load a color image
image = io.imread('tokyo.jpg')

# Obtain the segmentation with 1600 regions
segments = slic(image, n_segments=100)

# Put segments on top of original image to compare. 
# The color of each segment is the average color of pixels in that segment
segmented_image = label2rgb(segments, image, kind="avg")

plt.imshow(segmented_image)
````
