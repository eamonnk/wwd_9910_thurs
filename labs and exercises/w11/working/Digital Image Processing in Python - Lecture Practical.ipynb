### Task 4

Resize the tokyo image to 343 pixels width while preserving the its aspect ratio.


````python
from skimage import io, transform
import matplotlib.pyplot as plt

# Load an image
image = io.imread('tokyo.jpg')

# Resize preserving aspect ratio
resize_factor = 4

# create a new tuple
# we are re-sizing the image
# original length divided by 4 then sue transfform package to recall the resize methid... 
new_shape = (image.shape[0] / resize_factor, image.shape[1] / resize_factor)
resized_image = transform.resize(image, new_shape)

# can see axis belwo that its re-sized . one way to do it

# Display the resized image
plt.imshow(resized_image)
````

````python
# Task 4
# re-size
# keep same asp[ect ratio and set to new size
# 343 pixels

# Load an image
image = io.imread('tokyo.jpg')

# Define the new width you want
new_width = 343

# Calculate the new height to maintain the aspect ratio
original_width, original_height = image.shape[0], image.shape[1]
new_height = int((new_width / original_width) * original_height)

resized_image = transform.resize(image, (new_width, new_height))

# Display the resized image
plt.imshow(resized_image) 
````

````python
from skimage import io
import matplotlib.pyplot as plt
from skimage.color import rgb2gray
import numpy as np



# Load a color image
image = io.imread('https://i.ibb.co/h041g0p/Red-Cat-8727.jpg')

# Make the image grayscale using rgb2gray
image_gray = rgb2gray(image)
plt.hist(image_gray.ravel(), bins=256)
plt.show()

mask_lower_pixel_values = image_gray < 0.1
mask_higher_pixel_values = image_gray >= 2.0
#print(mask)
image_gray[mask_lower_pixel_values] = 0
image_gray[mask_higher_pixel_values] = 1
plt.imshow(image_gray, cmap='gray')
````
