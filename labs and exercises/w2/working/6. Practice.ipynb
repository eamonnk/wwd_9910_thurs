31. Count each of the items in this list. Output a dictionary where the key is the list-item and the value is the number of occurrences in the list

```python
items_to_count = ['a', 'b', 'b', 'c', 'd', 'f', 'a', 'b']


item_dict = {}


for item in items_to_count:
    if item in item_dict:
        item_dict[item] += 1
    else:
        item_dict[item] = 1

print(item_dict)
```
